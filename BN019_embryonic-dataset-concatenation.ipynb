# Load Datasets

## Morarach E15.5

```python
morarach_e15_ad = sc.read_h5ad('preprocessed_datasets/morarach_E15.h5ad')
```

```python
morarach_e15_ad.X = morarach_e15_ad.layers['raw_counts']
```

```python
morarach_e15_ad.obs['barcode'] = morarach_e15_ad.obs_names
```

```python
morarach_e15_ad.obs['age'] = 'E15.5'
morarach_e15_ad.obs['dataset'] = 'Morarach_E15.5'
```

```python
morarach_e15_ad.obs['sample'] = 'cell'
```

```python
morarach_e15_ad.obs = morarach_e15_ad.obs[['barcode', 'age', 'dataset', 'sample']]
```

```python
morarach_e15_ad.obs.head()
```

```python
morarach_e15_ad.var['gene_name'] = morarach_e15_ad.var_names
```

```python
morarach_e15_ad.var = morarach_e15_ad.var[['gene_name']]
```

```python
morarach_e15_ad.var.head()
```

```python
for key in list(morarach_e15_ad.layers.keys()):
    if key not in ['raw_counts', 'norm_counts', 'log_counts']:
        del morarach_e15_ad.layers[key]
```

```python
morarach_e15_ad = ad.AnnData(X = morarach_e15_ad.X,
                             obs = morarach_e15_ad.obs.copy(),
                             var = morarach_e15_ad.var.copy(),
                             layers = morarach_e15_ad.layers.copy()
                            )
```

```python
morarach_e15_ad
```
